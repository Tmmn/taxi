# Taxi Simulation - Batch Run

Generates config files and runs a batch of simulations

Usage:
1. Run *Generate configs* cell to create new config files
2. Run *Move configs* cell to collect the generated files into a subfolder
3. Edit `CONFIG_FOLDER` in the *Batch run* cell to point to that subfolder, then run it
4. Zip results

## Generate config files

Available modes: `sweep`, `passenger_fairness`, `region_pref`, `distance_pref`, `passenger_pref`, `two_sided`, `safety_objective`, `simple`

See [README.md](/README.md) for the full list of arguments per mode

```python
# Example: generate passenger_pref configs based on the big_city_base config for 5 simulated days
!python generate_configs.py passenger_pref big_city_base.conf 5 12
```

## Move generated configs to a batch folder

Set `GLOB_PATTERN` to match the files you just generated (based on the base config name and days used above) and `TARGET_FOLDER` to your desired batch subfolder

```python
import glob, shutil, os

# CHANGE THESE BEFORE RUN
GLOB_PATTERN = "configs/big_city_base_days_*.conf"
TARGET_FOLDER = "configs/passenger_pref"

os.makedirs(TARGET_FOLDER, exist_ok=True)
moved = 0
for f in glob.glob(GLOB_PATTERN):
    shutil.move(f, os.path.join(TARGET_FOLDER, os.path.basename(f)))
    moved += 1
print(f"Moved {moved} config(s) to {TARGET_FOLDER}/")
```

## Batch run

Set `CONFIG_FOLDER` to the directory that contains the `.conf` files you want to run.
Results are written to `results/<subfolder>/` and per-run logs to `batch_run_logs/`

```python
CONFIG_FOLDER = "configs/passenger_fairness"  # CHANGE THIS BEFORE RUN

!python batch_run.py {CONFIG_FOLDER}
```

## Zip results

```python
import shutil

shutil.make_archive("results", "zip", "results")
print("results.zip created")
shutil.make_archive("batch_run_logs", "zip", "batch_run_logs")
print("batch_run_logs.zip created")
shutil.make_archive("configs", "zip", "configs")
print("configs.zip created")
```
